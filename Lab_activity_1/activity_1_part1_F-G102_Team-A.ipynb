```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Spotify dataset

df = pd.read_csv('Spotify.csv')
df.head()
```

```python
years = df['year'].value_counts()
print("Data is recorded between the", years)
```

```python
counts = df['artist'].value_counts()
print("Number of Artists: ", len(counts))
print("--- All Artists & Song Counts ---")
print(counts)
```

```python
top_10 = counts.head(10)

print("--- Top 10 Artists & Song Counts ---")
print(top_10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 7))

ax = sns.barplot(
    x=top_10.index, 
    y=top_10.values, 
    palette='viridis', 
    hue=top_10.index, 
    legend=False
)
plt.title('Top 10 Artists with the most number of songs (2010-2011)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Artist Name', fontsize=12)
plt.ylabel('Number of Songs', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                fontsize=11, color='black', 
                xytext=(0, 5), 
                textcoords='offset points')

plt.tight_layout()

plt.show()
```
